```python
import chromadb
from sentence_transformers import SentenceTransformer

# Connect to your existing ChromaDB database
client = chromadb.PersistentClient(path="./chroma_db")
collection = client.get_or_create_collection(name="onboarding_faq")

# Load your embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

print(f"Connected to ChromaDB! Total chunks stored: {collection.count()}")

# Test a search query
question = "What is the onboarding policy?"
q_embedding = embedding_model.encode([question]).tolist()

results = collection.query(
    query_embeddings=q_embedding,
    n_results=2
)

print(f"\n--- Top Results for: '{question}' ---")
for i, doc in enumerate(results['documents'][0]):
    print(f"\n[Result {i+1}]\n{doc}")
```
